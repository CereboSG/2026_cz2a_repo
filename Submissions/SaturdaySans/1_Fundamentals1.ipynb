# Primitive Data Types, operators and expressions

* ```
  Expression ::= <operand> |

# Problem 1: Enter a mark and convert marks to grade:
* \>= 70 -> A,
* 60-69 -> B,
* 55-59 -> C ,
* 50-54 -> D,
* <50 -> Fail

````python
## Code here
def convertGrade(score):
    if score >= 70: return "A"
    elif score >= 60: return "B"
    elif score >= 55: return "C"
    elif score >= 50: return "D"
    return "Fail"



mark = 58

while True:
    mark = input("Input your score: ")
    print(f"Your grade is: {convertgrade(mark)}

print(convertGrade(mark))

````

# Data Validation vs Verification

Data validation is the process of checking that data is sensible, reasonable, and meets specified rules or formats before it is processed.
*   Presence Check: did user key in anything?
*   Range Check: did user key in something within the required range
*   Length Check: size of the input (number of characters)
*   Format Check: does input fulfill required format
___
*   Check Digits(Parity/Checksum), Example last letter on your NRIC
    -  Discuss later


Validation checks the input against certain rules

Data verification is the process of checking that data has been accurately transferred or copied from one source to another without errors.
Verification compares your input against something that you trust (trusted source)
*   visual check
*   double entry
