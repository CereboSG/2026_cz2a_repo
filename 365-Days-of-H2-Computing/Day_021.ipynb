## Exercise D21.1

Write a function that takes in a list of real numbers and returns the minimum value from that list. You may not use the built-in `min()` and `sorted()` functions, and you may not use the `list.sort()` method.

>```python
>your_function([1,2,3])
>1
>```

````python
#YOUR CODE HERE
import random

def your_function(lst):
    min = lst[0]
    for i in lst:
        if i < min: min = i
    return min
your_function([random.randint(1,100) for _ in range(5)])
````

````python
#DO NOT DELETE THE IMPORT
#you can find the square root of a number x by using the expression math.sqrt(x)
import math

#YOUR CODE HERE
def dot_product(vect1, vect2):
    return sum(x*y for x,y in zip(vect1, vect2))
def cross_product(vect1, vect2):
    return (vect1[1]*vect2[2] - vect1[2]*vect2[1], vect1[2]*vect2[0] - vect1[0]*vect2[2], vect1[0]*vect2[1] - vect1[1]*vect2[0])
def addition(vect1, vect2):
    return tuple(x+y for x,y in zip(vect1, vect2))
def subtraction(vect1, vect2):
    return tuple(x-y for x,y in zip(vect1, vect2))
def neg(vect1):
    return tuple(-x for x in vect1)
def mag(vect1):
    return math.sqrt(sum(x**2 for x in vect1))
def unit(vect1):
    magnitude = mag(vect1)
    return tuple(x/magnitude for x in vect1) if magnitude else (0,0,0)
````

## Exercise D21.5 Length of Projection of a Vector onto another Vector

Let $\mathbf{a}$, $\mathbf{b}$ be vectors and $\theta$ is the acute angle between the vectors.  The **length of projection** of $\mathbf{a}$ onto $\mathbf{b}$ is defined as $$\left|\mathbf{a}\right|\cos\theta=\frac{\left|\mathbf{a}\cdot\mathbf{b}\right|}{\left|\mathbf{b}\right|}=\left|\mathbf{a}\cdot\mathbf{\hat{b}}\right|.$$

Write a function named `length_of_proj` that:
- takes in 2 tuples `(x1,y1,z1)`, `(x2,y2,z2)` as parameters. Obviously, these tuples represent 3D vectors (not points!).
- return the length of projection of the first vector on the 2nd vector in the parameter.

Example interaction: 

>```python
>a = (2, 1, 2)
>b = (-2, -2, 4)
>length_of_proj(a, b)
>0.666666667
>length_of_proj(b, a)
>0.40824829
>```

````python
#YOUR CODE HERE
def length_of_projection(a,b):
    if mag(b)==0: return 0
    return abs(dot_product(a, b)) / mag(b)

a = (2, 1, 2)
b = (-2, -2, 4)
length_of_projection(b,a)
#order
````
